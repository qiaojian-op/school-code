```python
import mathFunctions
print("using dot notation")
print("sum: ",mathFunctions.summ(5,3))
print("sub: ",mathFunctions.sub(5,3))
print("Mul: ",mathFunctions.mul(5,3))
print("Fac: ",mathFunctions.factorial(5))

from mathFunctions import summ,mul
print("using import statement")
print("sum: ",summ(5,3))
print("Mul: ",mul(5,3))

import mathFunctions as mf
print("using alias")
print("sum: ",mf.summ(5,3))
print("sub: ",mf.sub(5,3))
print("Mul: ",mf.mul(5,3))
print("Fac: ",mf.factorial(5))
```
